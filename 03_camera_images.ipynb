```python
import cv2
from IPython.display import display, clear_output
from PIL import Image
import time

cap = cv2.VideoCapture(0)  # oder dein Gerät z. B. '/dev/video0'

for i in range(50):
    ret, frame = cap.read()
    if not ret:
        break

    # OpenCV -> RGB
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    img = Image.fromarray(frame_rgb)

    # Anzeige im Notebook
    clear_output(wait=True)
    display(img)

    time.sleep(0.1)  # etwas Verzögerung für Sichtbarkeit

cap.release()
```
