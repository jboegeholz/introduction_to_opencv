# Introduction to OpenCV

```python
import cv2 
import matplotlib.pyplot as plt 
import numpy as np

img = cv2.imread("doc_brown.png")
plt.imshow(img) 
```

```python
rgb_image = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) 
plt.imshow(rgb_image) 
```

```python
img.shape
```

## Brigthness

```python
brightness = np.zeros(rgb_image.shape, dtype="uint8") + 30 
brighter_image = cv2.add(rgb_image, brightness)
plt.imshow(brighter_image) 
```

## Adding noise

```python
noise = np.random.randint(0, 100, size=rgb_image.shape, dtype="uint8") 
noisy_image = cv2.add(rgb_image, noise)
plt.imshow(noisy_image)
```

## Smoothing aka Blurring

### Standard Blur

```python
blurred_image = cv2.blur(noisy_image, (3, 3))
plt.imshow(blurred_image)
```

#### different kernel sizes

```python
kernel_sizes = list(range(3, 10, 2))  # z. B. [3, 5, 7, 9]
n = len(kernel_sizes)

fig, axes = plt.subplots(1, n, figsize=(4 * n, 4))  # 1 Zeile, n Spalten

for ax, k in zip(axes, kernel_sizes):
    blurred = cv2.blur(noisy_image, (k, k))

    ax.imshow(blurred)
    ax.axis('off')
    ax.set_title(f"Kernel: {k}×{k}")

plt.tight_layout()
plt.show()
    
```

### Gaussian Blur

```python
image_with_gaussian_blur = cv2.GaussianBlur(noisy_image, ksize=(3, 3), sigmaX=5)
plt.imshow(image_with_gaussian_blur)
```

```python
kernel_sizes = list(range(3, 10, 2))  # z. B. [3, 5, 7, 9]
n = len(kernel_sizes)

fig, axes = plt.subplots(1, n, figsize=(4 * n, 4))  # 1 Zeile, n Spalten

for ax, k in zip(axes, kernel_sizes):
    blurred = cv2.GaussianBlur(noisy_image, (k, k), sigmaX=5)

    ax.imshow(blurred)
    ax.axis('off')
    ax.set_title(f"Kernel: {k}×{k}")

plt.tight_layout()
plt.show()
```

## Grayscale conversion

```python
gray_image = cv2.cvtColor(rgb_image, cv2.COLOR_BGR2GRAY) 
plt.imshow(gray_image, cmap='gray')
```

## Inversion

```python
imagem = cv2.bitwise_not(rgb_image)
plt.imshow(imagem)
```
